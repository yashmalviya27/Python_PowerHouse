<h2><b>Here we can chande the indexing</b></h2>

```python
data1 = {'PRODUCT NAME':['PRODUCT1', 'PRODUCT2', 'PRODUCT3'],
         'PRICE':[2000, 30000, 1233]}
index = ['a', 'b', 'c']
```

```python
df_1 = pd.DataFrame(data1, index = index)
df_1
```

This is the other method to display the Data Frame

```python
another_method = [{'PRODUCT': 'PRODUCT A', 'PRICE': 2000},
                 {'PRODUCT': 'PRODUCT A', 'PRICE': 2000}]
```
